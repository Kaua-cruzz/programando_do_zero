```python
notas = (6.7,8.0,8.9,5.8)

def calcular_media(notas):
    media = sum(notas) / len(notas)
    return media

media_calculada = calcular_media(notas)
arredondar_media = lambda media: round(media,2)
media_final = arredondar_media(media_calculada)

if media_final >= 7:
    situacao = 'aprovado'
else:
    situacao = 'reprovado'
print(f'Notas: {notas}')
print(f'Média: {media_final}')
print(situacao)
```

```python
def calcular_desconto(valor, percentual):
  if percentual < 0 or percentual > 100:
    return None
  desconto = valor * (percentual/100)
  return valor - desconto

def registrar_venda(produto,valor_final):
  print(f"venda registrada: {produto} po R${valor_final:.2f}")

arredondar = lambda v: round(v,2)

total_vendas = 0
qntd_vendas = 0

while True:
  print("---- minha loja ----")
  print("1 - registrar venda")
  print("2 - ver relatorio do dia")
  print("3 - sair")
  opcao = input("Escolha uma opção: ")

  if opcao == "3": # Changed from "0" to "3" for consistency with menu
    print("encerrando o caixa")
    break

  elif opcao == "1": # Opcao should be string from input
    produto = input("nome do produto: ")
    valor = float(input("valor do produto: "))
    percentual = float(input("desconto (%): "))

    valor_final = calcular_desconto(valor,percentual)

    if valor_final is None:
        print("desconto invalido: use um valor entre 0 e 100")
    else:
        valor_final = arredondar(valor_final)
        registrar_venda(produto, valor_final)
        total_vendas = total_vendas + valor_final
        qntd_vendas = qntd_vendas + 1

  elif opcao == "2":
    print(f"vendas de hoje:  {qntd_vendas}")
    print(f"total faturado: R$ {arredondar(total_vendas)}")

  else:
    print("opcao invalida. tente de novo")
    continue
```

```python
class veiculo:
  def __init__(self,marca,modelo,ano):
    self.marca = marca
    self.modelo = modelo
    self.ano = ano
    self.velocidade = 0
  def acelerar(self,valor):
    self.velocidade += valor
  def frear(self,valor):
    self.velocidade -= valor

class carro(veiculo):
  def __init__(self,marca,modelo,ano,potencia):
    super().__init__(marca,modelo,ano)
    self.potencia = potencia
  def acelerar(self, valor):
    self.velocidade += valor * self.potencia
  # Adicionado o método status para a classe carro
  def status (self):
    return f"marca: {self.marca}, modelo: {self.modelo}, ano: {self.ano}, velocidade: {self.velocidade}, potencia: {self.potencia}"

class bicicleta (veiculo):
  def __init__(self,marca,modelo,ano,tipo):
    super().__init__(marca,modelo,ano)
    self.tipo = tipo
  def status (self):
    return f"marca: {self.marca}, modelo: {self.modelo}, ano: {self.ano}, velocidade: {self.velocidade}, tipo: {self.tipo}"

carro1 = carro("toyota","corolla",2023,2.0)
bicicleta1 = bicicleta("treak", "mountain bike", 2021, "mtb")

carro1.acelerar(30)

print("status do carro: ")
# Chamando o método status na instância carro1
print(carro1.status())
print("\nstatus da bicicleta: ")
# Chamando o método status na instância bicicleta1
print(bicicleta1.status())
```

```python
import matplotlib.pyplot as plt

class produto:
  def __init__(self,nome,preco,categoria,estoque):
    self.nome = nome
    self.preco = preco
    self.categoria = categoria
    self.estoque = estoque

  def __str__(self):
    return (f"{self.nome} | R${self.preco:.2f} | categoria: {self.categoria} | estoque: {self.estoque}")

catalogo = []
categorias = []

def adicionar_produtos(nome, preco, categoria, estoque):
  novo_produto = produto(nome, preco, categoria, estoque)
  catalogo.append(novo_produto)
  categorias.append(categoria)
  print(f"produto '{nome}' adicionado ao catalogo")

def remover_produto(nome_produto):
  global catalogo, categorias
  produto_encontrado = None
  for p in catalogo:
    if p.nome == nome_produto:
      produto_encontrado = p
      break

  if produto_encontrado:
    catalogo.remove(produto_encontrado)
    categorias.remove(produto_encontrado.categoria) # Remove a categoria também
    print(f"Produto '{nome_produto}' removido do catálogo.")
  else:
    print(f"Produto '{nome_produto}' não encontrado no catálogo.")

def listar_catalogo():
  print("\n----catalogo de produtos da loja -----: ")
  if not catalogo:
    print("O catálogo está vazio.")
    return
  for produto in catalogo:
    print(produto)

adicionar_produtos("notebook",  3500.00, "informatica", 15)
adicionar_produtos("fone bluetooth", 170.00, "acessorio", 20)
adicionar_produtos("teclado", 150.00, "informatica", 39)
adicionar_produtos("cadeira", 3700.00, "movel", 7)

print()
listar_catalogo()
```

```python
categorias_unicas =  list(set(categorias))
categorias_unicas.sort()

contagem = [categorias.count(cat) for cat in categorias_unicas]

plt.bar(categorias_unicas, contagem)
plt.xlabel('categorias')
plt.ylabel('quantidade')
plt.title('quantidade de produtos por categoria')
plt.show()
```

```python
#tabela com CRUD completo
import sqlite3 as sql

conn = sql.connect("contatos.db")
cursor = conn.cursor()

# Apaga a tabela se ela já existir para evitar duplicatas em execuções repetidas
cursor.execute("DROP TABLE IF EXISTS contatos;")

# Cria a tabela de contatos com uma restrição UNIQUE para o email
cursor.execute("""
CREATE TABLE contatos (
    id INTEGER PRIMARY KEY,
    nome TEXT NOT NULL,
    telefone TEXT NOT NULL,
    email TEXT UNIQUE
);
""")
conn.commit()
conn.close() # Fecha a conexão após a criação da tabela
print("Tabela de contatos criada com sucesso e limpa para novas inserções.")

# Conecta novamente para inserir dados
conn = sql.connect("contatos.db")
cursor = conn.cursor()

dados_exemplos = [
    ("João", "joao@gmail.com", "123-456-7890"),
    ("Maria", "maria@gmail.com", "987-654-3210"),
    ("Carlos", "carlos@gmail.com", "555-555-5555")
]
cursor.executemany("INSERT INTO Contatos (nome, email, telefone) VALUES (?, ?, ?)", dados_exemplos)
conn.commit()
conn.close() # Fecha a conexão após a inserção dos dados
print("Dados de exemplo inseridos com sucesso.")

# Conecta novamente para ler os dados
conn = sql.connect("contatos.db")
cursor = conn.cursor()
cursor.execute("SELECT * FROM Contatos")
contatos = cursor.fetchall()
conn.close() # Fecha a conexão após a leitura
print("\nContatos após inserção:")
for contato in contatos:
  print(contato)

# Conecta novamente para atualizar
conn = sql.connect("contatos.db")
cursor = conn.cursor()
cursor.execute("UPDATE Contatos SET telefone = '111-111-1111' WHERE id = 1")
conn.commit()
conn.close() # Fecha a conexão após a atualização
print("\nProduto atualizado com sucesso.")

# Conecta novamente para deletar
conn = sql.connect("contatos.db")
cursor = conn.cursor()
cursor.execute("DELETE FROM Contatos WHERE id = ?", (2,))
conn.commit()
conn.close() # Fecha a conexão após a deleção
print("\nProduto deletado com sucesso.")

# Conecta novamente para a leitura final
conn = sql.connect("contatos.db")
cursor = conn.cursor()
cursor.execute("SELECT * FROM Contatos")
contatos = cursor.fetchall()
conn.close() # Fecha a conexão após a leitura final
print("\nContatos após atualização e deleção:")
for contato in contatos:
  print(contato)
```

```python
import kivy
kivy.require('1.0.0') # Replace with your Kivy version requirement if needed

!pip install kivy
!pip install kivymd

from kivy.lang import Builder

from kivy.uix.boxlayout import BoxLayout

from kivy.uix.gridlayout import GridLayout

from kivymd.app import MDApp

from kivymd.uix.button import MDRaisedButton

from kivymd.uix.textfield import MDTextField

from kivy.metrics import dp

KV = '''

<CalculatorApp>:

orientation: 'vertical'

MDTextField:

id: input_field

hint_text: "Insira um número"

helper_text_mode: "on_focus"

input_filter: "float"

GridLayout:

cols: 4

spacing: dp(10)

MDRaisedButton:

text: "1"

on_press: app.on_number_press(1)

MDRaisedButton:

text: "2"

on_press: app.on_number_press(2)

MDRaisedButton:

text: "3"

on_press: app.on_number_press(3)

MDRaisedButton:

text: "+"

on_press: app.on_operator_press("+")

MDRaisedButton:

text: "4"

on_press: app.on_number_press(4)

MDRaisedButton:

text: "5"

on_press: app.on_number_press(5)

MDRaisedButton:

text: "6"

on_press: app.on_number_press(6)

MDRaisedButton:

text: "-"

on_press: app.on_operator_press("-")

MDRaisedButton:

text: "7"

on_press: app.on_number_press(7)

MDRaisedButton:

text: "8"

on_press: app.on_number_press(8)

MDRaisedButton:

text: "9"

on_press: app.on_number_press(9)

MDRaisedButton:

text: "*"

on_press: app.on_operator_press("*")

MDRaisedButton:

text: "C"

on_press: app.clear_input()

MDRaisedButton:

text: "0"

on_press: app.on_number_press(0)

MDRaisedButton:

text: "="

on_press: app.calculate_result()

MDRaisedButton:

text: "/"

on_press: app.on_operator_press("/")

'''

class CalculatorApp(BoxLayout):

    def on_number_press(self, number):

        current_text = self.ids.input_field.text

        new_text = f"{current_text}{number}"

        self.ids.input_field.text = new_text

    def on_operator_press(self, operator):

        current_text = self.ids.input_field.text

        new_text = f"{current_text} {operator} "

        self.ids.input_field.text = new_text

    def clear_input(self):

        self.ids.input_field.text = ""

    def calculate_result(self):

        try:

            result = eval(self.ids.input_field.text)

            self.ids.input_field.text = str(result)

        except Exception as e:

            self.ids.input_field.text = "Error"


class CalculatorMDApp(MDApp):

    def build(self):

        return Builder.load_string(KV)

    def on_number_press(self, number):

        self.root.on_number_press(number)

    def on_operator_press(self, operator):

        self.root.on_operator_press(operator)

    def clear_input(self):

        self.root.clear_input()

    def calculate_result(self):

        self.root.calculate_result()

if __name__ == '__main__':

    CalculatorMDApp().run()
```
